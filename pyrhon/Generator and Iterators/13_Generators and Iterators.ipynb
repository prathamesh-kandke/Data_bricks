### Iterators
    -It is an object that can be used to traverse a sequence of values one at a time.
    -This are implemented using the iter() and next() method, which allows you to retreive the 
     next value in the sequence.

```python
lst=[11,22,33,44,55,66,77]
for num in lst:
    print(num)
it = iter(lst)
print(it)  
```

```python
it = iter(lst)
print(it)
```

```python
next(it)
```

```python
next(it)   #If it reaches last value then after that it will give error(stopIteration)
```

```python
it.__next__()  #this is another method 
```

### Generators
    Generator-function:
        -It is defined like a normal function but whenever it needs to generate a value , it does 
         with the 'yield' keyword rather than return.
        -If the body of a def conatains 'yield' , the function automatically becomes generator function.
    Generator Object:
        -Generator function returns a generator object.
        -Generator object are used by calling next() method on the generator object.

```python
#generator function
def g_func():
    yield 1
    yield 2
    yield 3
g_obj = g_func()  #stored in variable or object
next(g_obj)
```

```python
next(g_obj)
```

```python
next(g_obj)
```

```python
next(g_obj)
```

```python
def count_up_to(n):
    count=1
    while count <= n:
        yield count
        count+=1
count_up_to(12)
```

```python
cnt = count_up_to(12)
next(cnt)
```

```python
next(cnt)  #It will show up to 12
```
