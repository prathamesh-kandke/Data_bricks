## Inheritance :
   - The class from which properties are being derived is called as base/parent/superclass.
   - The class which derives the property is called as derived class/child/subclass.
   - In place of writing same code again and again we simply inherit the properties of one class into another class.
   
    Syntax:
        class parent_class:
            pass
        
        class child_class(parent_class)
            pass

```python
# create the parent class attributes are : name, gender, age
# create child class : city, department

class parent:
    def __init__(self, name, gender, age):
        self.name=name
        self.age=age
        self.gender=gender
        #print(self.name,self.age,self,gender)

class child(parent):
    def __init__(self, city, dept, name, gender, age):
        parent.__init__(self, name, gender, age)    
        self.city=city
        self.dept=dept
```

```python
c1=child('Pune', "Sales", "abc","M",19)
```

```python
c1.name, c1.gender
```

```python
class car:
    def __init__(self, wheel, color, engine, door):
        self.wheel=wheel
        self.color=color
        self.engine=engine
        self.door=door
        
    def driving(self):
        print("car is ready for driving")
```

```python
class nexon(car):
    def __init__(self, sunroof, fuel_type, wheel, color, engine, door):
        car.__init__(self, wheel, color, engine, door)
        self.sunroof=sunroof
        self.fuel_type=fuel_type
        
```

```python
n1=nexon("yes", "diesel", "alloy", "black", "1.5 ltr", "4")
```

```python
n1.engine
```

```python
n1.sunroof
```

```python
n1.driving()
```

### Multiple Inheritance:

```python
class parent_01:
    def __init__(self,name,gender,age):
        self.name=name
        self.gender=gender
        self.age=age
```

```python
class parnet_02:
    def __init__(self,city,dept):
        self.city=city
        self.dept=dept
```

```python
class child(parent_01, parnet_02):
    def __init__(self, name, gender, age, city, dept, sal):
        parent_01.__init__(self,name,gender,age)
        parnet_02.__init__(self,city,dept)
        self.sal=sal
```

```python
c1=child('Prathm',"M",20,"Pune","Marketing",25000)
```

```python
c1.name
```

```python
c1.sal
```

```python
c1.dept
```

### Multilevel Inheritance:
    crete the parent class having attribute name, gender, age
    create the another parent class inherting the properties of above class and p2 class having attribute city, dept
    create child class inheriting the p2 class and having attribute sal in it

```python
class p1:
    def __init__(self, name, gender, age):
        self.name=name
        self.gender=gender
        self.age=age
        
class p2(p1):
    def __init__(self,name, gender, age, city, dept):
        p1.__init__(self, name, gender, age)
        self.city=city
        self.dept=dept
```

```python
class child_1(p2):
    def __init__(self,name, gender, age, city, dept, sal):
        p2.__init__(self,name, gender, age, city, dept)
        self.sal=sal
```

```python
ch1=child_1("abc", "F", 25, "Mumbai", "HR",25000)
```

```python
ch1.sal
```

```python
ch1.dept
```

```python
ch1.age
```

### Hierarchical Inheritance :
    # create the parent class name as tata having attributes color, seats, engine, wheels
    # create the child 1 class name as nexon, inheriting the propertites of tata and nexon properties are fuel type
    # create the child 2 class name as altroz inherting properties of tata and altroz properties are sunroof 

```python
class tata:
    def __init__(self, color, seats, engine, wheels):
        self.color=color
        self.seats=seats
        self.engine=engine
        self.wheels=wheels
        
class nexon(tata):
    def __init__(self, color, seats, engine, wheels, fuel_type):
        tata.__init__(self, color, seats, engine, wheels)
        self.fuel_type=fuel_type
        
class altroz(tata):
    def __init__(self, color, seats, engine, wheels, sunroof):
        tata.__init__(self, color, seats, engine, wheels)
        self.sunroof=sunroof
```

```python
car1=nexon("Black", 5, "1.5 ltr", "alloy", "Diesel")
```

```python
car1.fuel_type
```

```python
car1.color
```

```python
car2=altroz("Yellow", 4, "2.0 ltr", "steel", "yes")
```

```python
car2.sunroof
```

```python
car2.engine
```

## Static Method :
   - The method that don't use the self parameter
   - Static method works at class level 
   - @staticmethod is used to create the static method 

```python
class employee:
    def __init__(self, eid, ename, sal, desig):
        self.eid=eid
        self.ename=ename
        self.sal=sal
        self.desig=desig
    
    @staticmethod    
    def display():
        print("Hello Everyone")
```

```python
employee.display()
```

```python
e1=employee(121,"abc",25000,"HR")
```

```python
e1.display()
```
