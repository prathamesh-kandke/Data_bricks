## Abstraction :
   - It hides the internal details and exposes only the necessary information or functionality
   - Abstraction is achieved by creating abstract class and abstract method  
   - In python abstraction is achieved by using abc(abstract base class) module.

```python
class circle:
    def __init__(self, r):
        self.r=r
    def area(self):
        print("Area of circle is :",3.14*self.r*self.r)
```

```python
c1=circle(5)
c1.area()
```

```python
class project():
    def __init__(self,a):
        self.a=a
    def phase_I(self):
        pass
    def phase_II(self):
        pass
```

```python
p1=project(10)
p1.a
```

```python
from abc import ABC, abstractmethod #abstract base class

class project_01(ABC):
    def __init__(self, a):
        self.a=a
    
    @abstractmethod
    def phase_1(self):
        pass
    #@abstractmethod
    def phase_2(self):
        pass
```

```python
p2=project_01(15)
```

```python
class application(project_01):
    def __init__(self, a):
        self.a=a
    
    def phase_1(self):
        print("Our team has started working on phase 1")
    # def phase_2(self):
    #     pass
```

```python
a=application(15)
a.phase_1()
print(a.a)
```
