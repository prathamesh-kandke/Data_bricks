## Polymorphism with class methods

```python
#create the circle class take radius as attribute and create the methods as area to print area of circle
# create the rectangle class take attribute as length & breadth and create the method name as area to print area of rectangle

class circle:
    def __init__(self, radius):
        self.radius=radius
    def area(self):
        print("Area of circle is :",3.14*self.radius*self.radius)

class rectangle:
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self):
        return self.l*self.b
```

```python
c1=circle(11)
r1=rectangle(10,2)
```

```python
c1.area()
```

```python
r1.area()
```

### Method Overriding
   - In this ths specific implementation of the method is present by the parent class and also it is present by the child class.
   - It is used to change the behaviour of the exising method and there is need for at least 2 class for method overloading.
   - In method overriding inheritance is always required as it is done between parent class and child class.

```python
class shape:
    def __init__(self,name):
        self.name=name
    def fact(self):
        print("I am 2D shape")
```

```python
s1=shape(rectangle)
```

```python
s1.fact()
```

```python
class circle(shape):
    def __init__(self,name,radius):
        shape.__init__(self,name)
        self.radius=radius
    def fact(self):
        print("I am 3D shape")
```

```python
c1=circle('circle',15)
```

```python
c1.fact()
```
