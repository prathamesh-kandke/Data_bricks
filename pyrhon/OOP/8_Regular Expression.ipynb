## Regular Expression :
   - It is a sequence of character used to check whether the pattern in text/string is avaiable or not.
   - To use regular expression we need to import it.

```python
import re
```

### Function of regular expression

#### 1. findall() :
   - Return the list containing all the matches 
   
    Syntax :
        re.findall('pattern', string_name)

```python
str1="We are learning Advance Python and advance excel with Power bi"

re.findall("[a-i]", str1)
```

### Pattern :

    '[]' --> Set of characters
    '[a-z]' ----> a,b,c,d,e,f,.....,z
    '[A-Z]' ----> A,B,C.......,Z
    '{ }' ----> Number of occurances    
    

    ### special characters
    \d 
    

```python
str2="Data Analyst, Data Engineer 878 112 1552"
re.findall("[A-Z]",str2)
```

```python
re.findall("[0-9]{3}",str2)
```

```python
re.findall("\d{3}",str2)
```

```python
re.findall("\w{3}",str2)
```
