## Encapsulation :
   - It refers to wrapping up the data under the single unit.
   - It is protective shield that prevents data from being accessed by  the code outside the shield.

```python
class student:
    def __init__(self,name, age):
        self.name=name #public attribute
        self.age=age
```

```python
s1=student("Amit",27)
```

```python
s1.age
```

```python
# access private attribute by creating methods
class bank:
    def __init__(self, accno, password):
        self.__accno=accno # private attribute
        self.__password=password

    def get_password(self): #private method
        print("Your password is :",self.__password)
```

```python
class bank:
    def __init__(self, accno, password):
        self.__accno=accno # private attribute
        self.__password=password

    def __get_password(self): #private method
        print("Your password is :",self.__password)
        
    def access_method(self):
        self.__get_password()
```

```python
b3=bank(12546,6666)
```

```python
b3.access_method()
```

```python
#create the bank class and create the private attribute as account and balance
# create the debit method to debit the particular amount
# create the credit method to credit the particular amount
# create the seperate method to access the private attribute i.e balance

class bank:
    def __init__(self,accno,bal):
        self.__accno=accno
        self.__bal=bal
    
    def debit(self,total):
        total = self.__bal - total
        print(f'Rs {self.__bal-total} is debited from your account')
        print("your balance is",total)
    
    def credit(self,total):
        total = self.__bal + total
        print(f'Rs {total-self.__bal} is credited from your account')
        print("your balance is",total)
        
    def show_balance(self):
        return self.__bal
        
```

```python
bank1=bank(1351,1000)
```

```python
bank1.debit(200)
```

```python
bank1.credit(300)
```

```python
bank1.show_balance()
```
