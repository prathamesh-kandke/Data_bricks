### Self :
    When we use the self: releated to attribute, method
    self is a current object which is present inside the class.

```python
class bank:
    def __init__(self, acc, passwrd):
        self.account=acc
        self.password=passwrd
    
    def show_info(self):
        print("Account no :",self.account)
        print("Password :",self.password)
```

```python
b1=bank(121,"123")
b1.show_info()
```

```python
#create the employee class and take the emp_id, emp_desig, department, as a argument
#create one display method to show only id and department

class employee:
    def __init__(self,emp_id,designation,department):
        self.emp_id=emp_id
        self.designation=designation
        self.department=department
    def display(self):
        print("Employee ID :",self.emp_id)
        print("Employee Department :",self.department)
```

```python
e1=employee(121,"HR","Web Development")
e1.display()
```

```python
#Cretae the student class that takes the name and marks of 3 subject as arguments
#create the method to print the average of 3 subject

class student:
    def __init__(self, name, marks):
        self.name=name
        self.marks=marks
    def output(self):
        print(self.marks)
        sum=0
        avg=0
        for i in self.marks:
            sum=sum+i
            avg=sum/len(self.marks)
        print("Average marks of Students :",avg)
```

```python
s1=student("Srushti",[99,88,87])
s1.output()
```

```python
# create the account class with 2 attribute balance, acc_no
# create the method for debit, credit and printing the balance

class account:
    def __init__(self, acc_no, bal):
        self.acc_no = acc_no
        self.bal = bal
        
    def credit(self,final):
        self.bal += final
        print(f'{final} amount credited successfully')
        print(f'Your balance is {self.bal}')
    
    def debit(self,final):
        if(self.bal>final):
            self.bal -= final
            print(f'{final} amount debited successfully')
            print(f'Your balance is {self.bal}')
        else:
            print("Insufficient Balance")
        
    def show_bal(self):
        print(f'Your balance is {self.bal}')
```

```python
a1=account(101,5000)
```

```python
a1.credit(200)
```

```python
a1.show_bal()
```

```python
a1.debit(500)
```

```python
a1.show_bal()
```
