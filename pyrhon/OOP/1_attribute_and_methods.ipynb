## Object Oriented Programming :
    An object oriented programming is used to design the program using class and object.

### Class :
    - Class is a blueprint or prototype for which object are been created.
    - It is a logical entity that have some specific attribute (variable) & methods (function).
    
    Syntax :
        class class_name:
            #attribute 
            #methods

```python
l1 = [1,2,3,4,5,6,7]
l1.append(10)
print(l1, type(l1))
```

```python
class Student:
    name="abc"
    age=30
    
    def greetings(self):
        print("hello everyone")
    
```

### Object :
    - It is a runtime entity which execute the class / represent the class.
    - Object is a entity that has some state and behaviour.
    - These variable/attributes are refered as properties of object & function/method are refered as behaviour of the object
        
    Syntax:
        object_name = class_name()

```python
s1 = Student()
```

```python
print(s1)
```
