```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_excel('pivot.xlsx')
df
```

```python
df1=df.pivot_table("Total",index="Party Name",aggfunc="sum").round()
df1
```

```python
df2=df.pivot_table(['Total','Qty'],index="Party Name",aggfunc={'Total':'mean','Qty':'sum'}).round()
df2
```

```python
df2.rename({'Qty':'Total Orders','Total':'Total Sales'} ,axis=1)
```

```python
df.pivot_table('Total',index="Party Name",columns='Item Name',aggfunc="sum")
```

```python
# top 2 item >> highest sale
# top 1 item >> order placed
# top salesman >> sales
```

```python
# top 2 item >> highest sale
df.pivot_table('Total', index='Item Name', aggfunc='sum').nlargest(2,'Total')
```

```python
# top 1 item >> order placed
df.pivot_table('Qty', index='Item Name', aggfunc='count').nlargest(2,'Qty')
```

```python
df.pivot_table('Bill No.', index='Item Name', aggfunc='count').nlargest(2,'Bill No.')
```

```python
# top salesman >> sales
df.pivot_table('Total', index='Sales Man', aggfunc='sum').sort_values('Total', ascending=False)
```
