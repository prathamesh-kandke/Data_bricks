## Tuple:

- When we want to store multiple value as a single entity we can use tuple.
- This value are enclosed in () brackets.
- Tuples are immutable datatype. 
- Heterogeneous elements are allowed.
- Insertion order is preserved.
- Indexing and slicing is supported in tuple.
- Tuples are faster than list.

```python
t1=(1,2,3,4,5)
type(t1)
```

```python
t1[2]
```

```python
t1[2] = 9
```

```python
t2=(1,2,3,4,'Prathmesh',3+5j,3.6)
print(t2)
```

```python
t3=(11,22,33,44,88,77,66,55,101)
print(t3[4:8])
```

```python
print(t3[-5:-1])
```

```python
lst=[1,2,3,'Mahesh',True,3.5,5+6j]
t4=tuple(lst)
print(t4)
```

```python
l1=[1,2,3,4]
l2=[10,11,12,13]

print(l1+l2)#when we add 2 list it will merge 
```

```python
t5=(1,2,3,4,5)
t6=(10,11,12,13)
print(t5 + t6)

print(t5)
print(t6)
```

```python
t5=(1,2,3,4,5)
t6=(10)

print(type(t5))
print(type(t6))

print(t5+t6)
```

## packing and unpacking of the tuple


```python
tup1= (10,20,30,40,50)

a,b,c,d,e = tup1 # packing of tuple

# unpacking of the tuple
print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
p,q,r = 'Mango','Apple','Cherry'
print(p)
```
