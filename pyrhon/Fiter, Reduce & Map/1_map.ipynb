## Map :
   - This map function in python is a build in function that allow you to apply a specfic function to every element of an iterable object (list, tuple, set, etc) and it return a new iterable object with the result
  
    Syntax :
        map(function , iterable)

```python
def square(n):
    return n*n
```

```python
square(2)
```

```python
l1=[1,2,3,4,5,6]

for i in l1:
    print(i**2)
```

```python
lst1=[1,2,3,4,5]
list(map(square,lst1))
```

```python
new = lambda l : list(map(square,lst1) )
```

```python
new([])
```

```python
list(map(lambda n : n*n , lst1))
```

```python
l1=[1,2,3,4]
l2=[10,20,30,40]

list(map(lambda a,b:a*b,l1,l2 ))
```

```python
def mul(p,q):
    return p*q

list(map(mul,l1,l2))
```
