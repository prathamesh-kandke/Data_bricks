## Reduce :
   - It is build in function in python functools module that apply a function of two arguments cumulatively of a sequence of element from left to right to reduce the sequence to a single value.
   
    Syntax :
        reduce(function,sequence)

```python
l1=[1,2,3,4,5]
```

```python
from functools import reduce

def mul(a,b):
    return a*b

output = reduce(mul,l1)
print(output)
```

```python
reduce(lambda p,q: p*q,l1)
```

```python
lst3=[11,22,33,11,22,33,66,77]

reduce(lambda a,b : a+b , lst3)
```

## Zip Function :
   - It is used to map a similar index of multiple sequence so that they can be used to generate new mapped sequence
    
    Synatax:
        zip(seq1,seq2)    

```python
lst4=['Phy', 'math', 'chem', 'bio', 'eng', 'hindi']
sub_marks=[74,88,73,86,77]

dict(zip(lst4,sub_marks))
```

```python
list(zip(lst4,sub_marks))
```
