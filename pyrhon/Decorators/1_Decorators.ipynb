## Closure :
   - When we write the function inside the function it is called as closures

```python
def main_wel():
    msg = 'Hello Everyone'
    def sub_wel():
        print("Hello Welcome to python training")
        print(msg)
    return sub_wel()
```

```python
main_wel()
```

```python
def main_wel(msg):
    def sub_wel():
        print("Hello Welcome to python training")
        print(msg)
    return sub_wel()

main_wel("hello")
```

```python
main_wel("Hiiii")
```

```python
main_wel("Good Evening")
```

```python
def main_wel(abc):
    def sub_wel():
        print("Welcome to python training")
        abc("Hello")
    return sub_wel()

main_wel(print)
```

```python
def main_wel(abc):
    def sub_wel():
        print("Welcome to python training")
        print(abc("Hello"))
    return sub_wel()


main_wel(len)
```

```python
main_wel(id)
```

```python
main_wel(type)
```

## Decorators :
   - It is a function which can take function as a argument and extends its functionality and return the modified function with extended functionality.

```python
def main_wel(func):
    def sub_wel():
        print("Hello Everyone")
        func()
    return sub_wel()

def greeting():
    print("We are learning decorators")
```

```python
main_wel(greeting)
```

```python
@main_wel
def greeting2():
    print("This is shortcut for decorators")
```
