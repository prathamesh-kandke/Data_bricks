## Continue :
    We can use continue to skip the current iteration and continue with next iteration.
        Syntax:
            for iterative_variable in sequence:
                if condition:
                    continue
                statement

```python
for num in range(1,11):
    print(num)
```

```python
for num in range(1,10):
    if num == 5:
        continue
    print(num)
```

```python
#print even number by using continue
for i in range(1,21):
    if i%2==0:
        print(i)
    else:
        continue
```

```python
# 7 table with continue
for i in range(1,71):
    if i % 7 == 0:
        print(i)
    else:
        continue
```

## break :
    We can use break inside loop to break the loop execution based on given condition
        Syntax:
            for iterative_variable in sequence:
                if condition:
                    break:
                statements


```python
for i in range(1,11):
    if i == 5:
        break
    print(i)
```

```python
lst=[1,3,7,9,4,6,2,5,11]
for i in lst:
    if i % 2 == 0:
        break
    print(i)
```

```python
# only print first even number from list
lst=[1,3,7,9,4,6,2,5,11]
for i in lst:
    if i % 2 == 0:
        print(i)
        break
```

## pass
    -In our programming syntactically if block is required which wont do anything then we use pass keyword to define empty block
    -The 'pass' statement is placeholder statement in python that does nothing. It is used as a placeholder when you need to implement
     a block of code later but don't have the code avaiable yet.

```python
for i in range(20):
    pass

print("Hello")
```

```python
lst=[128,75,145,25,228,575,632,557,700,32]
l2=[]
for i in lst:
    if i % 5 != 0:
        continue
    elif i >= 500:
        break
    else:
        l2.append(i)

print(l2)
```

```python
for i in range(10):
    if i == 3:
        continue
    if i ==5:
        break
    if i == 2:
        pass
    print(i)
```
