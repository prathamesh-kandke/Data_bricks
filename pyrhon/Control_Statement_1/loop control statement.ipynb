## Loop Control Statement
    
### For Loop:
    If we want to execute some action or statement number of times then we use this loop
    -Specially it is used with sequence(list, tuple, range, set ...)
    
    Syntax : 
        for iterative variable in sequence:
            # body of for loop
            statements 

```python
str1="Python"
print(str1[0])
print(str1[1])
print(str1[2])
print(str1[3])
print(str1[4])
print(str1[5])
```

```python
for i in str1:
    print(i)
```

```python
l1=[1,25,50,7,9,89.2,20]
for i in l1:
    print(i)
```

## For loop using range

    Syntax:
        for iterative variable in range(start_index, end_index(next_index), step_size):
        statements


```python
for i in range(1,11):
    print(i)
```

```python
for i in range(10,21):
    print(i)
```

```python
for i in range(10,0,-1):
    print(i)
```

```python
for i in range(1,11):
    #print(i**2)
    print(i*i)
```

```python
str1="data analyst"
for i in str1:
    print(i.upper())
```

```python
sum = 0
for i in range(1,11):
    if i % 2 == 0:
        sum += i
print("Sum of even number b/w 1 to 10 is",sum)
```

```python
n1=int(input("Enter 1 st number to start :"))
n2=int(input("Enter 2nd number to end :"))
total=0

for i in range(n1,n2+1):
    if i % 2 == 0:
        total += i
print(f'Sum of even number {n1} to {n2} is',total)
```

```python
total=0
total2=0
for i in range(1,11):
    if i % 2 ==0:
        total += i**2
        
    else:
        total2 += i**2
print("Even number Sum from 1 to 10 is",total)
print("Odd number Sum from 1 to 10 is",total2)
```

```python
#find sum of even number and sum of odd number b/w 1 to 40
odd=0
even=0
for i in range(1,40+1):
    if i % 2 == 0:
        even += i

    else:
        odd+=i
        
print("even",even)
print("odd",odd)
        
```

## Fibonacci Number

```python
a=0
b=1
c=0
n=int(input("Enter the number :"))
for i in range (0,n):
    c = a+b
    print(c)
    a = b
    b = c
    # b = a
    # a = c 
```

## While Loop
    
    if we want to execute block of code repetedly (number of times) until the condition condition becomes false
    
    Syntax:
        while condition:
            statements

```python
num = 0
while num <= 10:
    print(num)
    num += 1
    
```

```python
num = 10
while num >= 0:
    print(num)
    num -= 1
```

```python
num = 1
while num <= 5:
    print("Good afternoon")
    num += 1
```

```python
# count of even number from 1 to 20 by using while loop
num=1
cnt=0
while num<=20:
    if num%2==0:
        cnt +=1
    num += 1
   
print(cnt)
```

```python
num = 1
add = 0
while num<=10:
    add = add+num
    num+=1
print(add)
    
```

```python
num=int(input("Enter a number(0 to stop) :"))
total=0
while num != 0:
    total=total+num
    num=int(input("Enter a number(0 to stop) :"))
print(total)
```
