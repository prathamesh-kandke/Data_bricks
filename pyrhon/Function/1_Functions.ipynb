## Functions :
    - Function is group of releated statements that perform a specific task.
    - Function make our code more organized and managable.
    - Function helps to break our program into samller parts.
    - It makes our code reuseable.
    
        Syntax:
            ----for defining the function---
            def function_name(arguments/parameters) :
                statement1
                statement2
                statement3
                .
                .
                return statement
                
            ---for calling the function---
            function_name(argument/parameter)

```python
a=10
b=20

def add():
    print('Addition of two nos is', a+b)

add()
```

### Local Variables :
    - The varibles which are defined within the function or block.
    - Local variable only exist within the block they are declared in.

### Global Variable : 
    - The variables which are defined outside of all the functions.
    - Global variables remains in the memory for duration of the program.
