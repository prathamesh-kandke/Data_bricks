Machine Learning >> to give capability to a machine learn from past data for future prediction purpose or to make its own decision

Steps >>
1. Data Collection
2. Data Cleaning / Transformation
3. EDA (Data Analysis) -- Numerical reports/graphs
4. Model Selection
5. Model Training
6. Testing
7. Model Deployment

1. Data Collection >> pandas >> csv,sql,xsl,json


2. Data Transformation/ Cleaning >>
   - handle missing values
   - remove duplicate rows and columns
   - check data types
   - remove inconsistent data ..Gender>>M, F, Male
   - sampling (classification based) (balanced based) >> spam prediction 1,0
   - remove outliers >> z-score, IQR, ML algorithm (LOF, Isolation Forest, DBSCAN)

3. EDA >> data analysis.. graphs, stats/ numerical information
   - feature analysis
   - univariate, bivarates, multivariates 
   
4. Feature Selection .. to select best feature.. DT, corr, variance, filter, wrapper method    


5. Feature Engineering .. to create new feature by using existing feature


6. Seperate x, y
   - x >> input value
   - y >> output value


7. Encoding..categorical column to numerical.. x (no need on y)
   - Label encoding, onehot encoding, get_dummies

8. Scaling >> convert data in single scale
   - standardization (Standard normal distribution)
   - normalization (MIN MAX)
   - calculation time is reduced 
   - Uniform scaling

9. Train test split
   - training data ... X_Train, Y_Train
   - testing data...X_test, Y_test

10. Model Selection


11. Model Training(X_Train, Y_Train)


12. Model Testing(X_test, Y_test)... accuracy, performance of model


13. Deployment >> web app, Desktop application, DOCKER, AWS, GITHUB

```python
# IQR.. Inter Quartile Range
```

```python
df=pd.read_csv("weight-height.csv")
df
```

```python
Q3=df['Height'].quantile(0.75)
Q1=df['Height'].quantile(0.25)
```

```python
Q3, Q1
```

```python
IQR=Q3-Q1
IQR
```

```python
upper=Q3+1.5*IQR
lower=Q1-1.5*IQR
```

```python
upper, lower
```

```python
df_out=df[ (df['Height']>upper) | (df['Height']<lower) ]
df_out
```

```python
df_no_out=df[ (df['Height']<upper) & (df['Height']>lower) ]
```

- For tree based algo we don't need scaling


Advantages :-
- same scale
- Less calculation time
