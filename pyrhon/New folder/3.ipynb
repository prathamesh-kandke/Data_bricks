```python
df=pd.read_csv('car data.csv')
df
```

```python
df.shape
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True, ignore_index=True)
```

```python
df.shape, df.duplicated().sum()
```

```python
df['no_of_year']=2025-df['Year']
```

```python
df.drop('Year', inplace=True, axis=1)
```

```python
df.head()
```

```python
X=df.drop(['Car_Name', 'Selling_Price'], axis=1)
y=df['Selling_Price']
```

```python
X, y
```

```python
from sklearn.preprocessing import OneHotEncoder
oh=OneHotEncoder(sparse_output=False)
```

```python
df['Fuel_Type'].value_counts()
```

```python
df['Seller_Type'].value_counts()
```

```python
df['Transmission'].value_counts()
```

```python
en=oh.fit_transform(X[['Fuel_Type', 'Seller_Type', 'Transmission']])
en
```

```python
b=oh.get_feature_names_out()
b
```

```python
# X['x0_Fuel_Type', 'x1_Seller_Type', 'x2_Transmission'] = oh.fit_transform([['Fuel_Type', 'Seller_Type', 'Transmission']])
X[b]=en
```

```python
X[b]
```

```python
X.drop(['Seller_Type', 'Fuel_Type', 'Transmission'], inplace=True, axis=1)
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
```

```python
X.columns
```

```python
X[['no_of_year', 'Present_Price', 'Kms_Driven']]=sc.fit_transform(X[['no_of_year', 'Present_Price', 'Kms_Driven']])
```

```python
X
```

```python
X.describe()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(X, y, test_size=0.2, random_state=11)
```

```python
X_train.shape, X_test.shape
```

```python
y_train.shape, y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
```

```python
reg.fit(X_train, y_train)
```

```python
reg.score(X_test, y_test)
```

```python
y_pred=reg.predict(X_test)
y_pred
```

```python
pd.DataFrame({'y_test':y_test, 'y_pred':y_pred})
```
