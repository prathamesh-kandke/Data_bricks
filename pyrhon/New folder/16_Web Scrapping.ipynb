## Combining . with * and +:

- .* :- Matches any character (except newline) zero or more times. 
- This is often used to match everything within a line or between two delimiters.
- It's greedy by default, meaning it matches the longest possible string.
- .+ :- Matches any character (except newline) one or more times. This is similar to .* but 
- requires at least one character to be present. It's also greedy.


```python
import re

mystr='''The National Weather Service is an agency of the United States federal
government that is tasked with 12 13 14 providing weather forecasts, 
warnings of hazardous weather, and other weather-related products to organizations phone no 
12345-123 and 123 456 the public for the purposes of protection,orrrrr safety, 
and general to information'''

matches=re.findall(r'\d+', mystr)
print(matches)
```

```python
# greedy & non-greedy

import re

text = "<h1>Title</h1><p>Paragraph</p>"
pattern_greedy = "<.*>"  # Greedy: Matches the longest possible string
pattern_non_greedy = "<.*?>" # Non-Greedy: Matches the shortest possible string

matches_greedy = re.findall(pattern_greedy, text)
print(matches_greedy)  # Output: ['<h1>Title</h1><p>Paragraph</p>']

matches_non_greedy = re.findall(pattern_non_greedy, text)
print(matches_non_greedy)  # Output: ['<h1>', '</h1>', '<p>', '</p>']
```

```python
import re

# Example 1: Matching one or more characters between "a" and "b"
text1 = "ab"
text2 = "a1b"
text3 = "a12345b"
text4 = "acb"

pattern1 = "a.+b"

print(re.findall(pattern1, text1))  # Output: [] (No match, needs at least one character)
print(re.findall(pattern1, text2))  # Output: ['a1b']
print(re.findall(pattern1, text3))  # Output: ['a12345b']
print(re.findall(pattern1, text4))  # Output: ['acb']


# Example 2: Extracting everything after a specific word
text5 = "The quick brown fox jumps over the lazy dog."
pattern2 = "jumps.+"
print(re.findall(pattern2, text5))  # Output: ['jumps over the lazy dog.']
```

```python
# Example 3: Matching a string that starts with "start" and ends with "end"
text6 = "start something in between end"
text7 = "startend"
pattern3 = "start.+end"
print(re.findall(pattern3, text6))  # Output: ['start something in between end']
print(re.findall(pattern3, text7))  # Output: [ ]



# Example 4: Handling multiline strings (DOTALL flag)
text8 = """start
middle
end"""
pattern4 = "start.+end"
print(re.findall(pattern4, text8)) # Output: [] (doesnt match across lines by default)

pattern5 = "start.+end"
print(re.findall(pattern5, text8, re.DOTALL)) # Output: ['start\nmiddle\nend']
```

```python
# Example 5: Matching any characters including spaces
text9 = "Hello      World"
pattern6 = "Hello.+"
print(re.findall(pattern6, text9)) # Output: ['Hello      World']

# Example 6: Extracting part of a string between delimiters
text10 = "<tag>content</tag>"
pattern7 = "<tag>.+</tag>"
print(re.findall(pattern7, text10)) # Output: ['<tag>content</tag>']

# Example 7: Non-greedy matching with .+? (important difference)
text11 = "<tag>content1</tag><tag>content2</tag>"
pattern8 = "<tag>.+?</tag>" #Non-Greedy
pattern9 = "<tag>.+</tag>" #Greedy

print(re.findall(pattern8,text11)) # Output: ['<tag>content1</tag>', '<tag>content2</tag>']
print(re.findall(pattern9,text11)) # Output: ['<tag>content1</tag><tag>content2</tag>']
```

```python
countries_string = "India, Indonesia, Italia, America, china, IRELAND, Austria, INDIA"
pattern = r"I.*?A"  
matches = re.findall(pattern, countries_string, re.IGNORECASE)
print(matches)
```

```python
import re

countries_string = "India, Indonesia, Italia, America, china, IRELAND, Austria, INDIA"
list1 = countries_string.split(',')

pattern = r"^i.*a$"

for country in list1:
    country = country.strip()  # Remove leading/trailing spaces
    if re.findall(pattern, country, re.IGNORECASE):
        print(country)
```

```python
# I... fourth I... A
import re

countries_string = "India, Indonesia, Italia, America, china, IRELAND, Austria, INDIA"
list1 = countries_string.split(',')

pattern = r"^i..i*a$"

for country in list1:
    country = country.strip()  # Remove leading/trailing spaces
    if re.findall(pattern, country, re.IGNORECASE):
        print(country)
```
