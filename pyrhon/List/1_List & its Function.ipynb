## List :-
   - If we want to represent group of individual elements as single entity then we use list.
   - It is collection of values, enclosed in [ ].
   - Insertion order is preserved.
   - Duplicate value are allowed.
   - Hetrogenous elements are allowed.
   - It is mutable datatype (we can add, delete, update the elements).
   - List supports Indexing & Slicing.

```python
l1=[1,2,3,4,5]
type(l1)
```

```python
lst1=[1,2,3,4,5,"Mahesh"]
print(lst1)
```

```python
lst1=[1,2,3,4,4,4,"Mahesh"]
print(lst1)
```

```python
lst2=[1,2,3,4,5,'mahesh',3.5,(6+8j),True]
print(lst2)
```

## Function of list
### 1. Append:-
    We can use append() to add element at the end of list.
    
    Syntax:
        list_name.append(value)

```python
lst6 = [1,2,3,4,5]
lst6.append(6)
print(lst6)
```

```python
l2=[1,2,3,4,5]
l2.append([6,7])
print(l2)
```

```python
lst3=[12,15,13,16,26,22,21,37,39,36,34]
even_list=[]
odd_list=[]

for i in lst3:
    if i % 2 == 0:
        even_list.append(i)
    else:
        odd_list.append(i)
print("Even Numbers -",even_list)
print("Odd Numbers -",odd_list)
```

### 2. insert()
    -to insert element at specific index location in the list
    
    Syntax:
        list_name.insert(index,value)

```python
lst5=[10,20,30,40]

lst5.insert(2,60)
print(lst5)
```

```python
lst5=[10,20,30,40,50,60,70]

lst5.insert(-3,11)
print(lst5)
```

### 3. extend()-
    - to add the all elements in one sequence of the list
    
    Syntax:
        list_name.extend(sequence)

```python
lst6=[10,20,30,40,50,60,70]
lst7=[11,22,33,44]

lst6.extend(lst7)
print(lst6)
```

```python
lst6=[10,20,30]
str1='Python'

lst6.extend(str1)
print(lst6)

#difference between append and extend
```

### 4.remove()-
    - it will remove the specific element from the list.
    - if element is present multiple time in list then only first occurance will be removed.
    - if element not found in list then it will give error. 
    
    Syntax:
        list_name.remove(element)  

```python
lst6=[10,20,30,40,50,60,70]

lst6.remove(30)
print(lst6)
```

```python
lst6=[10,20,30,40,50,60,20,100,25,20]
lst6.remove(20)
print(lst6)
```

### 5. pop-
    - it will remove the element of the specific index location.
    - it is the only function that return the removed value.
    - it will remove the last index element if index number is not mentioned.

    Synatx:
        list_name.pop(index)
        

```python
lst6 = [10,20,30,40,50,60,70,20,30,40,100,20]
lst6.pop(4)
```

```python
lst6.pop()
```

### 6. clear-
    - it will removw all the elements from given list

    Syntax:
        list_name.clear()

```python
lst6 = [10,20,30,40,50,60,70,20,30,40,100,20]
lst6.clear()
print(lst6)
```

```python
# Shallow Copy
list1 = [10,20,30,40]
list2 = list1
list2[2]="Python"
print(list1)
print(list2)
print(id(list1), id(list2))
```

### 7. copy()-
    - it will copy all the elements from one list to another list.
    
    Syntax:
        list_name2 = list_name1.copy() 

```python
# Deep copy
list3=[10,20,30,40]
list4=list3.copy()
list4[2]="hii"
print(list3)
print(list4)
print(id(list3), id(list4))
```

### 8. reverse()- 
    - we can use reverse the order of element in the list
    
    Syntax:
        list_name.reverse()

```python
lst5=[1,2,3,4]
lst5.reverse()
print(lst5)
```

### 9. sort()-
    - it wil do ascending as well as desecding sorting
    - it is used to sort the element in the list
    
    Syntax:
        list_name.sort(reverse=false) 

```python
lst6=[1,5,3,8,9,34,11,22]
lst6.sort()
print(lst6)
```

```python
lst6=[1,5,3,8,9,34,11,22]
lst6.sort()
print(lst6)

lst6.sort(reverse=True)
print(lst6)
```

```python
lst3=[12,15,13,12,16,26,22,16,21,37,15,26]
unique_value=[]

for i in lst3:
    if i not in unique_value:
        unique_value.append(i)
        
print(unique_value)
```

```python
lst3=[12,15,13,12,16,26,22,16,21,37,12,15,26]
unique_value=[]
duplicate_value=[]

for i in lst3:
    if i not in unique_value:
        unique_value.append(i)
    else:
        if i not in duplicate_value:
            duplicate_value.append(i)
        
print(unique_value)
print(duplicate_value)
    
```
