## Numpy :-
   - Numerical Python
   - can create multidimensional array and perform operation on that

### Array:-
   - Datatype in python
   - container datatype which can store multiple values
   - support indexing
   - mutable >> change values
   - homogenous >> same data type
   - array is faster than list
   - numpy lib >> c, python
   - 1D array >> list >> vector >> column >> series
   - 2D array >> list in list >> table >> matrix >> dataframe
   - 3D array >> list in list in list >> panel
