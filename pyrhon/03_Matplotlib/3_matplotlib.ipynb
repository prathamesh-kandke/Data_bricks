## Matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd
```

### Line chart
   - pattern , relationship, trend between variables

```python
days=[1,2,3,4,5]
temp=[34, 37, 39, 42, 46]
```

```python
plt.plot(days, temp)
plt.show()
```

```python
plt.plot(days, temp, color='red')
plt.show()
```

```python
plt.plot(days, temp, color='green', marker='.', linestyle='--', label='Pune City')
plt.xlabel('Days')
plt.ylabel("Temp")
plt.title("Pune City")
plt.xticks(range(1,6))
plt.yticks(range(34,51))
plt.legend() 
plt.show()
```

```python
plt.figure(figsize=(13,13))
plt.plot(days, temp, color='green', marker='.', linestyle='--', label='Pune City')
plt.xlabel('Days')
plt.ylabel("Temp")
plt.title("Pune City")
plt.xticks(range(1,6))
plt.yticks(range(34,51))
plt.legend() # indicator 
plt.show()
```

```python
#mutiple line chart in same plot
```

```python
days=[1,2,3,4,5]
ptemp=[34,37,39,42,46]
ntemp=[35,37,39,41,44]
```

```python
plt.plot(days, ptemp, color='green', marker='.', linestyle='--', label='Pune City')
plt.plot(days, ntemp, color='blue', marker='*', linestyle='--', label='Nashik City')
plt.xlabel('Days', fontsize=12)
plt.ylabel("Temp")
plt.title("Pune City & Nashik City")
plt.xticks(range(1,6))
plt.yticks(range(34,51))
plt.legend(loc=4) # indicator 
plt.show()
```

### Histogram
   - frequency distribution over the range of interval
   - univariate analysis

```python
df=pd.read_csv('iris.csv')
```

```python
plt.hist(df['sepal_length'], color='r', bins=[4,5,6,7,8])
plt.show()
```

```python
df['sepal_length'].plot(kind='kde')
```
