## 1. Union:
    - The union operation on the set return a new set that contains all unique element from both the set
    - We can pass number of sequence,  seperated by comma
    
    Syntax:
        set1.union(set2,list1,tup...)

```python
set1={1,2,3,4,5}
set2={3,4,5,6,7}
lst1=[10,5,8,9]

set1.union(set2,lst1)
```

## Intersection:
    - The inntersection operation takes the two set as input and return a new set that contain only the elements that are present in both set.
    - In other words, the intersection operation return the common elements between two sets.
    
    Syntax:
        set1.intersection(set2)

```python
set3={1,2,3,4,5,6}
set4={5,6,7,8,9,10}
lst1=[5,6,7,8,22]

set3.intersection(set4,lst1)
```

## Intersection_update() :
    -Method is used to update set with the intersection of itself and another set
    
    Syntax:
        set1.intersection_update(set2)

```python
set3={1,2,3,4,5,6}
set4={5,6,7,8,9,10}

set3.intersection_update(set4)

print(set3)
```

```python
set5={11,22,33,44}
set6={52,22,44,86,94,10}
set5.intersection_update(set6)
print(set5)
```

## Difference()
    -Method returns a new set that contains all the elements that are in 'set1' but not in 'set2'
    
    Syntax:
        set1.difference(set2)

```python
set1={1,2,3,4}
set2={3,4,5,6}
set1.difference(set2)
```

```python
set2.difference(set1)
```

## Difference_update() :
    -Method is used to update a set with the difference of itself and the another set.
       
    Syntax:
        set1.difference_update(set2)

```python
set3={1,2,3,4,5,6}
set4={5,6,7,8,9,10}
set3.difference_update(set4)
print(set3)
```
